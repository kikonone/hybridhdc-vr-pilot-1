## Runtime and execution-state summary

```python
efficiency_root = PHASE_DIR / 'results/efficiency'
efficiency_files = list(efficiency_root.rglob('*.json'))
records = [(path, json.loads(path.read_text(encoding='utf-8'))) for path in efficiency_files]
preprocessing_unique = {}
encoding_unique = {}
for path, item in records:
    relative = path.relative_to(efficiency_root)
    modality, task, filename = relative.parts
    fold = filename.split('_seed_')[0]
    seed = filename.split('_seed_')[1].split('_metrics')[0].split('_efficiency')[0]
    preprocessing_unique.setdefault((modality, fold), item['preprocessing_seconds'])
    encoding_unique.setdefault((modality, fold, seed), item['encoding_seconds'])
runtime = {'efficiency_records': len(records), 'shared_preprocessing_fits': len(preprocessing_unique), 'shared_encoder_runs': len(encoding_unique), 'preprocessing_seconds_deduplicated': sum(preprocessing_unique.values()), 'encoding_seconds_deduplicated': sum(encoding_unique.values()), 'task_training_seconds': sum(item['training_seconds'] for _, item in records), 'task_inference_seconds': sum(item['inference_seconds'] for _, item in records)}
print(json.dumps({'runtime_summary': runtime, 'execution_manifest': execution_manifest}, indent=2)); assert len(records) == 250 and len(preprocessing_unique) == 25 and len(encoding_unique) == 125
```
